```python
import os
import numpy as np
from PIL import Image
from matplotlib.colors import Normalize

import matplotlib.pyplot as plt

# List of image files produced by the earlier cells
image_files = [
    "../assets/red_blood_cell.jpg",
    "../results/example_results/Red_Blood_Cell_SimpleMicroscope.jpg",
    "../results/example_results/Red_Blood_Cell_ThickLensMicroscope.jpg",
    "../results/example_results/Red_Blood_Cell_MicroscopeObjective.jpg",
]

for path in image_files:
    if not os.path.exists(path):
        print(f"Skipping missing file: {path}")
        continue

    # Load image
    img = Image.open(path).convert("L")
    img_array = np.array(img)

    # Compute 2D FFT and shift it to center
    fft_img = np.fft.fft2(img_array)
    fft_shifted = np.fft.fftshift(fft_img)

    # Create a circular mask to remove low frequencies
    rows, cols = img_array.shape
    crow, ccol = rows // 2, cols // 2
    radius = min(rows, cols) // 100

    y, x = np.ogrid[:rows, :cols]
    mask = (x - ccol) ** 2 + (y - crow) ** 2 <= radius**2

    masked_fft = fft_shifted.copy()
    masked_fft[mask] = 0

    # Use log scale for better visualization
    fft_mag = np.log(1 + np.abs(masked_fft))

    # Display original image and Fourier transform side by side
    fig, axes = plt.subplots(1, 2, figsize=(12, 6))
    axes[0].imshow(img_array, cmap="gray")
    axes[0].set_title("Original Image")
    axes[0].axis("off")
    norm = Normalize(vmin=fft_mag.min(), vmax=fft_mag.max())
    sm = plt.cm.ScalarMappable(norm=norm, cmap="gray")
    sm.set_array([])
    fig.colorbar(sm, ax=axes[1], fraction=0.046, pad=0.04, label="Intensity")
    axes[1].imshow(fft_mag, cmap="gray")
    axes[1].set_title("FFT (Low Frequencies Removed)")
    axes[1].axis("off")

    plt.tight_layout()
    # Reconstruct image by inverse FFT of the masked spectrum
    ifft_shifted = np.fft.ifftshift(masked_fft)
    img_recon = np.fft.ifft2(ifft_shifted)
    img_recon = np.abs(img_recon)

    # Normalize to 0-255 and convert to uint8
    img_recon -= img_recon.min()
    img_recon = img_recon / (img_recon.max() + 1e-12) * 255
    img_recon = img_recon.astype(np.uint8)

    # Display reconstructed image
    fig2, ax2 = plt.subplots(figsize=(6, 6))
    ax2.imshow(img_recon, cmap="gray")
    ax2.set_title("Reconstructed Image (Low Freq Removed)")
    ax2.axis("off")
    plt.tight_layout()
    plt.show()
```
