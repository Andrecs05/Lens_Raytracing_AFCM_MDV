###Simple microscope

```python
SM_lens = ThinLens(25)

Simple_microscope = OpticalSystem(color=True)
Simple_microscope.add_element(SM_lens)

SM_img_path = "../assets/red_blood_cell.jpg"
SM_img = Image.open(SM_img_path)
SM_img_array = np.array(SM_img)
SM_object = Object(SM_img_array, distance=20, height=0.5)

A, B, C, D = Simple_microscope.MG.flatten()
d_img = -(A * SM_object.distance + B) / (C * SM_object.distance + D)

SM_image_plane_propagation = FreeSpace(d_img)
Simple_microscope.add_element(SM_image_plane_propagation)

pixel_size = abs(Simple_microscope.MG[0, 0]) * SM_object.height / SM_img_array.shape[0]  # Calculate pixel size based on magnification and object height
print(f"Calculated pixel size for the microscope: {pixel_size:.6f} mm")
#pixel_size = 0.001  # Example pixel size in mm
NA = 0.08 # Numerical Aperture of the microscope objective
pupil_radius = NA * SM_object.distance  # Calculate pupil radius based on NA and object distance
n_rays_per_pixel = 33  # Number of rays per pixel

SM_output_image_array = Simple_microscope.image_object(SM_object, pupil_radius, pixel_size, n_rays_per_pixel, interpolation=False)
SM_output_image = Image.fromarray(SM_output_image_array)
SM_output_image.save("../results/example_results/Red_Blood_Cell_SimpleMicroscope.jpg")
print("Image saved as Red_Blood_Cell_SimpleMicroscope.jpg")

SM_output_image_array_IP = Simple_microscope.image_object(SM_object, pupil_radius, pixel_size, n_rays_per_pixel, interpolation=True, min_intensity=0)
SM_output_image_IP = Image.fromarray(SM_output_image_array_IP)
SM_output_image_IP.save("../results/example_results/Red_Blood_Cell_SimpleMicroscope_Interpolated.jpg")
print("Interpolated image saved as Red_Blood_Cell_SimpleMicroscope_Interpolated.jpg")
```

###Microscope objective

```python
n_NBK7 = refractive_index(0.5876, 'NBK7')
n_NBAF10 = refractive_index(0.5876, 'NBAF10')

MO_lens1 = ThickLens(100000, -3.0, 3, n_NBK7)
MO_free_space1 = FreeSpace(0.3)
MO_lens2 = ThickLens(6, 15, 2, n_NBK7)
MO_free_space2 = FreeSpace(0.8)
MO_lens3 = ThickLens(10, -10, 2, n_NBAF10)

Microscope_objective = OpticalSystem(color=True)
Microscope_objective.add_element(MO_lens1)
Microscope_objective.add_element(MO_free_space1)
Microscope_objective.add_element(MO_lens2)
Microscope_objective.add_element(MO_free_space2)
Microscope_objective.add_element(MO_lens3)

MO_img_array = SM_img_array  # Using the same red blood cell image
MO_object = Object(MO_img_array, distance=0.9, height=0.5)

A, B, C, D = Microscope_objective.MR.flatten()
d_img = -(A * MO_object.distance + B) / (C * MO_object.distance + D)
print(f"Image distance for the microscope objective: {d_img:.2f} mm")
MO_image_plane_propagation = FreeSpace(d_img)
Microscope_objective.add_element(MO_image_plane_propagation)

pixel_size = abs(Microscope_objective.MR[0, 0]) * MO_object.height / MO_img_array.shape[0]  # Calculate pixel size based on magnification and object height
print(f"Calculated pixel size for the microscope objective: {pixel_size:.6f} mm")
NA = 0.65
pupil_radius = NA * MO_object.distance  # Calculate pupil radius based on NA and object distance
n_rays_per_pixel = 33  # Number of rays per pixel
print(f"Calculated pupil radius for the microscope objective: {pupil_radius:.6f} mm")

MO_output_image_array = Microscope_objective.image_object(MO_object, pupil_radius, pixel_size, n_rays_per_pixel, interpolation=False)
MO_output_image = Image.fromarray(MO_output_image_array)
MO_output_image.save("../results/example_results/Red_Blood_Cell_MicroscopeObjective.jpg")
print("Image saved as Red_Blood_Cell_MicroscopeObjective.jpg")

MO_output_image_array_IP = Microscope_objective.image_object(MO_object, pupil_radius, pixel_size, n_rays_per_pixel, interpolation=True, min_intensity=0)
MO_output_image_IP = Image.fromarray(MO_output_image_array_IP)
MO_output_image_IP.save("../results/example_results/Red_Blood_Cell_MicroscopeObjective_Interpolated.jpg")
print("Interpolated image saved as Red_Blood_Cell_MicroscopeObjective_Interpolated.jpg")
```
